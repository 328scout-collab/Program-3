## The Problem
The problem that you will be solving is called the **MAXIMALSUBSTRING** problem.

### MAXIMALSUBSTRING
* **Input:** A string $a$ of $n$ characters, and a string $b$ of $m$ characters
* **Output:** A string $s$ such that $s$ is a substring of both $a$ and $b$, and there is no string $s'$ where $|s'| > |s|$ and $s'$ is a substring of both $a$ and $b$

For example, if we are given the strings `"orange"` and `"banana"`, we would return `"an"`; for `"banana"` and `"apple"`, we return `"a"`; and for `"apple"` and `"pineapple"`, we return `"apple"`. For `"pineapple"` and `"orange"`, we could return either `"e"` or `"a"`.

There are $O(n^2)$ possible substrings in $a$ and (similarly) $O(m^2)$ in $b$. We can avoid some complexity by only checking substrings of the same length, but a brute-force algorithm will still check $O(n^2 m)$ pairs.

I strongly encourage you to use character operations rather than string operations—they will be easier to index usefully, and are no less efficient.

> **N.B.** String operations are not constant time. Comparing two strings takes $O(k)$ time, where $k$ is the string length. Similarly, concatenation $x + y$ takes $O(k_y)$ time (some implementations require $k_x + k_y$; we will assume ours is not one such). Slicing (getting a substring) needs time proportional to the length of the substring.

Your goal is to create a faster algorithm using a dynamic-programming strategy. To do this, you will first implement a backtracking (brute-force) strategy, then memoize it lazily while keeping the structure of the backtracking algorithm, then rebuild the dynamic-programming algorithm to eagerly fill the appropriate array.

## Approach

The core logic of our algorithm checks the common suffix length for all pairs of end indices for each string, then keeps track of the longest common suffix and returns that substring.

This works because every substring of $a$ and $b$ must terminates at some pair of indices $(i, j)$. Our algorithm computes the length of the longest common suffix ending at every pair $(i, j)$. Because every valid common substring corresponds to a common suffix for some pair of endpoints, tracking the maximum value across all $(i, j)$ guarantees that the globally longest common substring is identified without checking redundant slices.

We start with a backtracking approach that uses a recursive sub-algorithm to return the common suffix length for some pair of indices $(i, j)$. Then we loop through all pairs of indices and use that recursive function to calculate the common suffix for each pair while keeping track of the maximum.

Then we convert the algorithm into a dynamic-programming approach by using a 2D memo ($n \times m$) to keep track of the common suffix length for all pairs of end indices $(i, j)$ to avoid doing redundant computations for the same pairs of indices.

Finally, we convert the algorithm into an eager approach that builds the memo bottom up without using recursive calls since each element only depends on the row and column before it.

We can also further optimize this by only keeping a single array of length $min(n, m)$ instead of a $n \times m$ table. In order to do this we swap $a$ and $b$ if needed to make sure $b$ is the shorter string, then we iterate $i$ from $1$ to $n$ and $j$ from $m$ to $1$ to ensure that we don't overwrite the value from the previous row while only using a single array. Since the algorithm is symmetric, swapping $a$ and $b$ doesn't change the result.

## Pseudocode

### Backtracking

```text
Algorithm MAXIMAL-SUBSTRING-BACKTRACKING(a, b):
    n = length(a)
    m = length(b)

    if n == 0 or m == 0:
        return ""

    Function COMMON-SUFFIX-LEN(i, j):
        if i < 0 or j < 0:
            return 0

        if a[i] == b[j]:
            return 1 + COMMON-SUFFIX-LEN(i - 1, j - 1)
        else:
            return 0
    
    max_len = 0
    end_idx_a = -1

    for i from 0 to n - 1:
        for j from 0 to m - 1:
            curr_len = COMMON-SUFFIX-LEN(i, j)
            if curr_len > max_len:
                max_len = curr_len
                end_idx_a = i

    if max_len == 0:
        return ""

    start_idx_a = end_idx_a - max_len + 1
    return a[start_idx_a : end_idx_a + 1]
```

### Memoized

```text
Algorithm MAXIMAL-SUBSTRING-MEMOIZED(a, b):
    n = length(a)
    m = length(b)

    if n == 0 or m == 0:
        return ""

    memo = [[-1 for col in range(m)] for row in range(n)]

    Function COMMON-SUFFIX-LEN(i, j):
        if i < 0 or j < 0:
            return 0
        
        if memo[i][j] != -1:
            return memo[i][j]

        if a[i] == b[j]:
            memo[i][j] = 1 + COMMON-SUFFIX-LEN(i - 1, j - 1)
        else:
            memo[i][j] = 0
    
        return memo[i][j]
    
    max_len = 0
    end_idx_a = -1

    for i from 0 to n - 1:
        for j from 0 to m - 1:
            curr_len = COMMON-SUFFIX-LEN(i, j)
            if curr_len > max_len:
                max_len = curr_len
                end_idx_a = i

    if max_len == 0:
        return ""

    start_idx_a = end_idx_a - max_len + 1
    return a[start_idx_a : end_idx_a + 1]
```

### Eager

```text
Algorithm MAXIMAL-SUBSTRING-EAGER(a, b):
    n = length(a)
    m = length(b)

    if n == 0 or m == 0:
        return ""

    memo = [[0 for col in range(m + 1)] for row in range(n + 1)]
    
    max_len = 0
    end_idx_a = -1

    for i from 1 to n:
        for j from 1 to m:            
            if a[i - 1] == b[j - 1]:
                memo[i][j] = 1 + memo[i - 1][j - 1]
                if memo[i][j] > max_len:
                    max_len = memo[i][j]
                    end_idx_a = i - 1
            else:
                memo[i][j] = 0

    if max_len == 0:
        return ""

    start_idx_a = end_idx_a - max_len + 1
    return a[start_idx_a : end_idx_a + 1]
```

### Eager (Optimized)

```text
Algorithm MAXIMAL-SUBSTRING-EAGER-OPTIMIZED(a, b):
    if length(a) < length(b):
        c = a
        a = b
        b = c

    n = length(a)
    m = length(b)

    if n == 0 or m == 0:
        return ""

    memo = [0 for row in range(m + 1)]
    
    max_len = 0
    end_idx_a = -1

    for i from 1 to n:
        for j from m to 1:            
            if a[i - 1] == b[j - 1]:
                memo[j] = 1 + memo[j - 1]
                if memo[j] > max_len:
                    max_len = memo[j]
                    end_idx_a = i - 1
            else:
                memo[j] = 0

    if max_len == 0:
        return ""

    start_idx_a = end_idx_a - max_len + 1
    return a[start_idx_a : end_idx_a + 1]
```
